```python
import pandas as pd

df = pd.read_csv('../data/raw/hr_attrition.csv')
print('Dimensiones:', df.shape)
```

```python
df.info()
```

```python
constantes = [c for c in df.columns if df[c].nunique() == 1]
print('Columnas constantes:', constantes)
```

```python
df = df.drop(columns=constantes)
print('Dimensiones:', df.shape)
```

```python
print('Nulos totales:', df.isnull().sum().sum())
print('Filas duplicadas:', df.duplicated().sum())
print('EmployeeNumber duplicados:', df['EmployeeNumber'].duplicated().sum())
```

```python
print(df['Attrition'].value_counts())
print('Tasa de rotación:', round((df['Attrition'] == 'Yes').mean() * 100, 2), '%')
```

```python
rotacion = df.groupby('Department')['Attrition'].apply(lambda s: (s == 'Yes').mean() * 100)
print(rotacion.round(2).sort_values(ascending=False))
```

```python
rot_rol = df.groupby('JobRole')['Attrition'].apply(lambda s: (s == 'Yes').mean() * 100)
print(rot_rol.round(2).sort_values(ascending=False))
```

```python
df.groupby('Attrition')[['MonthlyIncome', 'Age', 'YearsAtCompany']].mean().round(1)
```

```python
df.groupby('OverTime')['Attrition'].apply(lambda s: (s == 'Yes').mean() * 100).round(2)
```

```python
df.groupby(['OverTime', 'Attrition'])['MonthlyIncome'].mean().round(0)
```

```python
df.groupby('JobSatisfaction')['Attrition'].apply(lambda s: (s == 'Yes').mean() * 100).round(2)
```

```python
df.groupby('YearsSinceLastPromotion')['Attrition'].apply(lambda s: (s == 'Yes').mean() * 100).round(2)
```

```python
df['YearsSinceLastPromotion'].value_counts().sort_index()
```

```python
df['Perfil_Riesgo'] = (
    (df['OverTime'] == 'Yes') & (df['JobSatisfaction'] <= 2)
).map({True: 'Horas extra + satisfacción baja', False: 'Otros'})

df['Perfil_Riesgo'].value_counts()
```

```python
perfil = df.groupby('Perfil_Riesgo')['Attrition'].agg(
    Empleados='size',
    Se_fueron=lambda s: (s == 'Yes').sum(),
)
perfil['Rotación (%)'] = (perfil['Se_fueron'] / perfil['Empleados'] * 100).round(2)
perfil
```

```python
factores = {
    'Hace horas extra': df['OverTime'] == 'Yes',
    'Satisfacción baja (1 o 2)': df['JobSatisfaction'] <= 2,
    'Salario menor a 5,000': df['MonthlyIncome'] < 5000,
}

total_salidas = (df['Attrition'] == 'Yes').sum()
filas = []
for nombre, mascara in factores.items():
    grupo = df[mascara]
    salidas = (grupo['Attrition'] == 'Yes').sum()
    filas.append({
        'Factor': nombre,
        'Empleados': len(grupo),
        'Se_fueron': salidas,
        'Rotación (%)': round(salidas / len(grupo) * 100, 2),
        '% de todas las salidas': round(salidas / total_salidas * 100, 1),
    })

pd.DataFrame(filas).set_index('Factor')
```

```python
pd.crosstab(df['BusinessTravel'], df['OverTime'], normalize='index').round(3) * 100
```

```python
df.groupby(['OverTime', 'BusinessTravel'])['Attrition'].apply(lambda s: (s == 'Yes').mean() * 100).round(2)
```

```python
df['Riesgo_Alto'] = (
    (df['OverTime'] == 'Yes') &
    (df['JobSatisfaction'] <= 2) &
    (df['BusinessTravel'] == 'Travel_Frequently')
)

riesgo = df.groupby('Riesgo_Alto')['Attrition'].agg(
    Empleados='size',
    Se_fueron=lambda s: (s == 'Yes').sum(),
)
riesgo['Rotación (%)'] = (riesgo['Se_fueron'] / riesgo['Empleados'] * 100).round(2)
riesgo
```
